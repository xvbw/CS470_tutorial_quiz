## 1. Select the class and classroom

- Course: enter `1` for CS470 AI or `2` for CS670 AI.
- Name: enter your full name (example: `Jon Doe`).
- UMB student ID: copy your own ID. `12345678` is only an example, not an ID to submit.
- Building: choose the number shown beside the correct building (for example, enter `3` to choose option 3).
- Room: copy it exactly from the UMB course listing in the format `W00-0000` (W, two digits, hyphen, four digits).

The notebook stops if a menu number is not valid, a required identity field is blank, or the room does not match the required format.


````python
from enum import IntEnum
import re

class Course(IntEnum):
    CS470_AI = 1
    CS670_AI = 2

class Building(IntEnum):
    CAMPUS_CENTER = 1
    EDWARD_M_KENNEDY_INSTITUTE = 2
    HEALEY_LIBRARY = 3
    INTEGRATED_SCIENCES_COMPLEX = 4
    MCCORMACK_HALL = 5
    NANTUCKET_FIELD_STATION = 6
    QUINN_ADMINISTRATION_BUILDING = 7
    MOTLEY_RESIDENCE_HALL_EAST = 8
    MOTLEY_RESIDENCE_HALL_WEST = 9
    SERVICE_AND_SUPPLY_BUILDING = 10
    UNIVERSITY_HALL = 11
    WHEATLEY_PETERS_HALL = 12
    CALF_PASTURE_PUMPING_STATION = 13
    CAMPUS_CENTER_LAWN = 14
    JFK_PRESIDENTIAL_LIBRARY_AND_MUSEUM = 15
    MASSACHUSETTS_ARCHIVES = 16

print("Course options:")
print("1. CS470 AI")
print("2. CS670 AI")
course_choice = int(input("Course number (1 = CS470 AI, 2 = CS670 AI): "))
assert course_choice in Course._value2member_map_, "Choose a listed course number."
COURSE_CODE = {1: "CS470 AI", 2: "CS670 AI"}[course_choice]
STUDENT_NAME = input("Full name (example: Jon Doe): ").strip()
STUDENT_ID = input("UMB student ID (enter your own; example only: 12345678): ").strip()
assert STUDENT_NAME, "Enter your name."
assert STUDENT_ID, "Enter your UMB student ID."

print("Building options:")
for building in Building:
    print(f"{building.value}. {building.name.replace(chr(95), chr(32)).title()}")
BUILDING_NUMBER = int(input("Building number (enter the number beside your building; e.g., 3): "))
assert BUILDING_NUMBER in Building._value2member_map_, "Choose a listed building number."
BUILDING = Building(BUILDING_NUMBER).name
ROOM = input("Room from UMB course listing (e.g., W00-0000): ").strip().upper()
assert re.fullmatch(r"W\d{2}-\d{4}", ROOM), "Enter the room in the required W00-0000 format."
CLASSROOM = f"{BUILDING} {ROOM}"
print("Selected class:", COURSE_CODE)
print("Selected building number:", BUILDING_NUMBER)
print("Selected room:", ROOM)
````

## 2. Print the local execution receipt

Start Jupyter from your activated Conda environment. This cell prints a stable device hash and a run-specific receipt. Recognized hosted services are reported when standard environment markers are present; no marker can prove who typed or assisted with the code.


````python
import hashlib
import os
import platform
from datetime import datetime, timezone

import torch
import torchvision

assert os.environ.get("CONDA_PREFIX"), "Activate your Conda environment before starting Jupyter."
host_facts = "|".join((platform.node(), platform.system(), platform.machine()))
device_stamp = hashlib.sha256(host_facts.encode("utf-8")).hexdigest()[:16]
run_time = datetime.now(timezone.utc).isoformat(timespec="seconds")
run_receipt = hashlib.sha256(f"{COURSE_CODE}|{CLASSROOM}|{device_stamp}|{run_time}".encode()).hexdigest()[:16]
known_hosts = {
    "Google Colab": "COLAB_RELEASE_TAG",
    "Kaggle": "KAGGLE_KERNEL_RUN_TYPE",
    "GitHub Actions": "GITHUB_ACTIONS",
    "GitHub Codespaces": "CODESPACES",
    "JupyterHub": "JUPYTERHUB_USER",
}
detected_hosts = [name for name, key in known_hosts.items() if os.environ.get(key)]
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Course code:", COURSE_CODE)
print("Student name:", STUDENT_NAME)
print("UMB student ID:", STUDENT_ID)
print("Classroom:", CLASSROOM)
print("UTC execution time:", run_time)
print("Device stamp:", device_stamp)
print("Run receipt:", run_receipt)
print("OS / architecture:", platform.system(), "/", platform.machine())
print("Conda active: yes")
print("PyTorch / torchvision:", torch.__version__, "/", torchvision.__version__)
print("Compute device:", device)
print("Recognized hosted-runtime indicators:", detected_hosts or "none detected")
````

## 3. Download the pretrained model and its required image transform


````python
from torchvision.models import ResNet18_Weights, resnet18

weights = ResNet18_Weights.DEFAULT
model = resnet18(weights=weights).to(device).eval()
preprocess = weights.transforms()
labels = weights.meta["categories"]
print("Pretrained weights:", weights)
````

## 4. Download and classify the sample photo

The sample photo and pretrained weights download automatically on first run. The model returns ImageNet classes; top-5 output is sufficient.


````python
from pathlib import Path
from urllib.request import urlretrieve
from PIL import Image

image_path = Path("data/dog.jpg")
image_path.parent.mkdir(parents=True, exist_ok=True)
if not image_path.exists():
    urlretrieve("https://github.com/pytorch/hub/raw/master/images/dog.jpg", image_path)
image = Image.open(image_path).convert("RGB")
batch = preprocess(image).unsqueeze(0).to(device)
with torch.no_grad():
    probabilities = model(batch).softmax(dim=1)[0]
top_probs, top_ids = probabilities.topk(5)
for rank, (prob, idx) in enumerate(zip(top_probs, top_ids), start=1):
    print(f"{rank}. {labels[int(idx)]}: {float(prob):.3f}")
````
